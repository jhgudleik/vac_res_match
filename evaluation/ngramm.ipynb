```python
import re

def text_preproc(txt):
  return ' '.join(re.sub(r'[^а-яА-Яa-zA-Z ]', '', txt).lower().strip().split())

def get_ngramm(n,txt):
  txt = txt.lower()
  res_text = []
  for w in list(txt.split()):
    for ng in range(len(w) - n + 1):
      res_text.append(w[ng:ng+n])
  return ' '.join(res_text)
```

```python
cv_processed = get_ngramm(3,text_preproc(cv))
vacancy1_processed = get_ngramm(3,text_preproc(vacancy1))
vacancy2_processed = get_ngramm(3,text_preproc(vacancy2))
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
cv_vec = model.encode(cv_processed).reshape(1, -1)
vacancy1_vec = model.encode(vacancy1_processed).reshape(1, -1)
vacancy2_vec = model.encode(vacancy2_processed).reshape(1, -1)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
similarity_score1 = cosine_similarity(cv_vec, vacancy1_vec)
similarity_score2 = cosine_similarity(cv_vec, vacancy2_vec)
```

```python
print(f"Cosine Similarity: {np.clip(similarity_score1[0][0], -1.0, 1.0)}")
print(f"Cosine Similarity: {np.clip(similarity_score2[0][0], -1.0, 1.0)}")
```
