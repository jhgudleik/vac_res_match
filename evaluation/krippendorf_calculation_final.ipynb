```python
import krippendorff
from sklearn.metrics import cohen_kappa_score
import numpy as np
import pandas as pd
```

```python
#Конвертация оценочных данных из имеющегося формата и требуемый для функции
def data_preproc(l1):
  l = []
  for r in list(range(0,30)):
    for v in list(range(0,5)):
      l.append(l1[r][v])
  return l

for k,v in auto_raters.items():
  auto_raters[k].append(data_preproc(v[0]))
for k,v in manual_raters.items():
  manual_raters[k].append(data_preproc(v[0]))
```

```python
def krippendorf_calculation(arr):
  alpha_nominal_custom = krippendorff.alpha(
    reliability_data=arr,
    value_domain=list(range(1,6)),
    level_of_measurement="ordinal"
)
  return alpha_nominal_custom
```

```python
#Вычисляем коэффициент криппендорфа для пары оценок
i=0
for km in manual_raters.keys():
  for ka in auto_raters.keys():
    auto_raters[ka].append(krippendorf_calculation([manual_raters[km][1],auto_raters[ka][1]]))
    auto_raters[ka].append(cohen_kappa_score(manual_raters[km][1],auto_raters[ka][1],weights='linear'))
```
