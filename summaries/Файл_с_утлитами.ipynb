```python
import pandas as pd
import json
from typing import List, Dict, Any
import os


def load_sample_vacancies() -> List[Dict[str, Any]]:
    """
    Загружает примеры вакансий для тестирования

    Returns:
        Список тестовых вакансий
    """
    vacancies = [
        {
            "id": "vac_001",
            "title": "Python Developer",
            "text": """We are looking for a Python Developer with at least 3 years of experience.

            Responsibilities:
            - Develop and maintain backend services using Python
            - Work with Django/Flask frameworks
            - Write clean, maintainable code and unit tests
            - Participate in code reviews
            - Collaborate with frontend developers

            Requirements:
            - 3+ years of experience with Python
            - Strong knowledge of Django or Flask
            - Experience with PostgreSQL, Redis
            - Familiarity with Docker and Kubernetes
            - Understanding of REST API design

            Nice to have:
            - Experience with React or Vue.js
            - Knowledge of AWS services
            - CI/CD pipeline experience

            We offer:
            - Competitive salary: $90,000 - $120,000 per year
            - Remote work options
            - Flexible working hours
            - Health insurance
            - Professional development budget""",
            "source": "sample"
        },
        {
            "id": "vac_002",
            "title": "Senior Java Backend Developer",
            "text": """Java Backend Developer needed for our microservices platform.

            Key Responsibilities:
            - Design and implement microservices using Java 11+
            - Develop RESTful APIs and event-driven services
            - Optimize application performance and scalability
            - Work with database design and optimization

            Required Skills:
            - 5+ years of Java development experience
            - Strong knowledge of Spring Boot, Spring Cloud
            - Experience with PostgreSQL and Kafka
            - Familiarity with Docker and Kubernetes
            - Understanding of distributed systems

            Preferred Qualifications:
            - Experience with Apache Kafka or RabbitMQ
            - Knowledge of NoSQL databases (MongoDB, Cassandra)
            - AWS or Azure cloud experience

            Benefits:
            - Competitive compensation package: $130,000 - $160,000
            - Stock options
            - Comprehensive health benefits
            - 401(k) matching
            - Remote-friendly culture""",
            "source": "sample"
        },
        {
            "id": "vac_003",
            "title": "React Frontend Developer",
            "text": """Join our team as a React Frontend Developer to build amazing user interfaces.

            What you'll do:
            - Develop responsive web applications using React 18
            - Implement state management with Redux or Context API
            - Write unit and integration tests with Jest
            - Collaborate with UX/UI designers
            - Optimize application performance

            Must have:
            - 2+ years of React experience
            - Proficiency in TypeScript
            - Strong CSS/HTML skills
            - Experience with modern frontend build tools
            - Understanding of responsive design

            Bonus points for:
            - Experience with Next.js or Gatsby
            - Knowledge of GraphQL
            - UI/UX design skills
            - Testing frameworks experience

            Perks:
            - Flexible remote work
            - Annual learning stipend: $5,000
            - Latest hardware
            - Team offsites
            - Wellness programs

            Location: Hybrid (New York)
            Salary: $85,000 - $110,000""",
            "source": "sample"
        },
        {
            "id": "vac_004",
            "title": "DevOps Engineer",
            "text": """DevOps Engineer to manage our cloud infrastructure and CI/CD pipelines.

            Responsibilities:
            - Design and implement CI/CD pipelines using Jenkins
            - Manage cloud infrastructure (AWS/Azure)
            - Implement monitoring and alerting systems with Prometheus/Grafana
            - Ensure system security and compliance
            - Automate deployment processes

            Requirements:
            - 3+ years of DevOps experience
            - Strong knowledge of AWS or Azure
            - Experience with Docker and Kubernetes
            - Proficiency in Terraform or CloudFormation
            - Knowledge of Linux administration

            Desired Skills:
            - Experience with Prometheus, Grafana
            - Knowledge of Ansible or Chef
            - Security certifications (AWS, Kubernetes)
            - Python or Go scripting experience

            What we provide:
            - Competitive salary + bonus: $120,000 - $150,000
            - Remote-first workplace
            - Cutting-edge technology stack
            - Conference attendance budget
            - Generous vacation policy

            Work type: Full-time, Remote""",
            "source": "sample"
        },
        {
            "id": "vac_005",
            "title": "Data Scientist",
            "text": """Data Scientist position for our machine learning team.

            Your role:
            - Develop and deploy machine learning models
            - Analyze large datasets and extract insights
            - Conduct A/B testing and statistical analysis
            - Collaborate with product and engineering teams
            - Present findings to stakeholders

            Required Qualifications:
            - Masters or PhD in Computer Science, Statistics, or related field
            - 3+ years of data science experience
            - Proficiency in Python and data science libraries (pandas, numpy, scikit-learn)
            - Strong SQL skills
            - Experience with ML frameworks (TensorFlow, PyTorch)

            Preferred Experience:
            - Experience with Apache Spark or Databricks
            - Knowledge of MLOps practices
            - Cloud platform experience (AWS SageMaker, Azure ML)
            - Big data technologies

            Company Benefits:
            - Equity compensation
            - Comprehensive medical/dental/vision
            - Parental leave
            - Home office setup allowance
            - Learning and development programs

            Location: San Francisco or Remote
            Salary Range: $140,000 - $180,000""",
            "source": "sample"
        }
    ]

    return vacancies


def load_vacancies_from_csv(filepath: str) -> List[Dict[str, Any]]:
    """
    Загружает вакансии из CSV файла

    Args:
        filepath: Путь к CSV файлу

    Returns:
        Список вакансий в формате для суммаризатора
    """
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Файл не найден: {filepath}")

    df = pd.read_csv(filepath)
    vacancies = []

    # Определяем колонки
    id_col = None
    title_col = None
    text_col = None

    # Ищем подходящие колонки
    for col in df.columns:
        col_lower = col.lower()
        if 'id' in col_lower:
            id_col = col
        elif 'title' in col_lower or 'name' in col_lower:
            title_col = col
        elif 'text' in col_lower or 'description' in col_lower or 'content' in col_lower:
            text_col = col

    # Если не нашли текстовую колонку, используем все колонки
    if text_col is None:
        text_col = df.columns[0]

    for idx, row in df.iterrows():
        vacancy = {
            "id": row[id_col] if id_col else f"vac_{idx+1:03d}",
            "title": row[title_col] if title_col else f"Vacancy {idx+1}",
            "text": str(row[text_col])
        }
        vacancies.append(vacancy)

    return vacancies


def load_vacancies_from_json(filepath: str) -> List[Dict[str, Any]]:
    """
    Загружает вакансии из JSON файла

    Args:
        filepath: Путь к JSON файлу

    Returns:
        Список вакансий
    """
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Файл не найден: {filepath}")

    with open(filepath, 'r', encoding='utf-8') as f:
        data = json.load(f)

    # Проверяем формат данных
    if isinstance(data, list):
        return data
    elif isinstance(data, dict):
        return [data]
    else:
        raise ValueError("Неверный формат JSON данных")


def print_summary_statistics(df: pd.DataFrame):
    """
    Выводит аналитическую статистику по результатам

    Args:
        df: DataFrame с результатами суммаризации
    """
    if df.empty:
        print("Нет данных для анализа")
        return

    print("\n" + "="*70)
    print("АНАЛИТИЧЕСКАЯ СТАТИСТИКА ПО ВАКАНСИЯМ")
    print("="*70)

    successful = df[df['confidence_score'] > 0.3]

    print(f"Всего вакансий: {len(df)}")
    print(f"Качественно обработано: {len(successful)} ({len(successful)/len(df)*100:.1f}%)")

    if len(successful) > 0:
        # Распределение по локациям
        locations = successful['location'].value_counts().head(5)
        print("\nТоп-5 локаций:")
        for loc, count in locations.items():
            if pd.notna(loc) and loc != "Not specified":
                print(f"  {loc}: {count}")

        # Распределение по типам работы
        work_types = successful['work_type'].value_counts()
        print("\nТипы работы:")
        for w_type, count in work_types.items():
            if pd.notna(w_type) and w_type != "Not specified":
                print(f"  {w_type}: {count}")

        # Диапазоны зарплат
        salary_stats = successful[sal := successful['salary_info'].str.contains('\$', na=False)]['salary_info']
        if len(salary_stats) > 0:
            print(f"\nВакансий с указанной зарплатой: {len(salary_stats)}")

            # Извлекаем числа из зарплат
            salaries = []
            for salary_str in salary_stats:
                numbers = re.findall(r'\$?(\d{1,3}(?:,\d{3})+)', str(salary_str))
                if numbers:
                    for num in numbers:
                        clean_num = int(num.replace(',', ''))
                        salaries.append(clean_num)

            if salaries:
                print(f"  Медианная зарплата: ${pd.Series(salaries).median():,.0f}")
                print(f"  Средняя зарплата: ${pd.Series(salaries).mean():,.0f}")

        # Время обработки
        avg_time = successful['processing_time'].mean()
        print(f"\nСреднее время обработки: {avg_time:.2f} секунд")

        # Сжатие текста
        avg_original = successful['source_text_length'].mean()
        avg_summary_len = successful['summary_english'].str.len().mean()
        compression = (1 - avg_summary_len / avg_original) * 100
        print(f"Среднее сжатие текста: {compression:.1f}%")

    print("="*70)


def export_to_excel(df: pd.DataFrame, filename: str = "vacancy_analysis.xlsx"):
    """
    Экспортирует результаты в Excel с несколькими листами

    Args:
        df: DataFrame с результатами
        filename: Имя файла Excel
    """
    try:
        with pd.ExcelWriter(filename, engine='openpyxl') as writer:
            # Основной лист
            df.to_excel(writer, sheet_name='Все вакансии', index=False)

            # Лист с высоким confidence
            high_conf = df[df['confidence_score'] > 0.7]
            if len(high_conf) > 0:
                high_conf.to_excel(writer, sheet_name='Высокая уверенность', index=False)

            # Лист с технологиями
            tech_data = []
            for idx, row in df.iterrows():
                if pd.notna(row['extracted_technologies']):
                    for tech in str(row['extracted_technologies']).split(','):
                        tech = tech.strip()
                        if tech:
                            tech_data.append({
                                'vacancy_id': row['vacancy_id'],
                                'vacancy_title': row['vacancy_title'],
                                'technology': tech,
                                'confidence': row['confidence_score']
                            })

            if tech_data:
                tech_df = pd.DataFrame(tech_data)
                tech_df.to_excel(writer, sheet_name='Технологии', index=False)

            # Лист со статистикой
            stats = {
                'Метрика': [
                    'Всего вакансий',
                    'Успешно обработано',
                    'Средняя уверенность',
                    'Средняя длина текста',
                    'Среднее время обработки',
                    'Вакансий с зарплатой',
                    'Удалённых вакансий',
                    'Гибридных вакансий'
                ],
                'Значение': [
                    len(df),
                    len(df[df['confidence_score'] > 0.3]),
                    df['confidence_score'].mean(),
                    df['source_text_length'].mean(),
                    df['processing_time'].mean(),
                    len(df[df['salary_info'].str.contains('\$', na=False)]),
                    len(df[df['location'] == 'Remote']),
                    len(df[df['location'] == 'Hybrid'])
                ]
            }
            stats_df = pd.DataFrame(stats)
            stats_df.to_excel(writer, sheet_name='Статистика', index=False)

        print(f"\n[ЭКСПОРТ] Результаты сохранены в Excel: {filename}")

    except Exception as e:
        print(f"[ОШИБКА] Не удалось сохранить в Excel: {e}")
```
