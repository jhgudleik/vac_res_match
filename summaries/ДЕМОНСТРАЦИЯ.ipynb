```python
"""
Демонстрация работы суммаризатора вакансий
"""

import pandas as pd
import sys
import os

# Добавляем путь для импорта
sys.path.append(os.path.dirname(os.path.abspath(__file__)))

from vacancy_summarizer import create_summarizer
from vacancy_data_utils import load_sample_vacancies, print_summary_statistics, export_to_excel


def demo_quick_summary():
    """Быстрая демонстрация работы"""
    print("БЫСТРАЯ ДЕМОНСТРАЦИЯ СУММАРИЗАЦИИ ВАКАНСИЙ")


    # Создаем суммаризатор
    summarizer = create_summarizer(verbose=True)

    # Загружаем примеры вакансий
    vacancies = load_sample_vacancies()

    # Обрабатываем и создаем CSV
    print(f"\nОбрабатываю {len(vacancies)} тестовых вакансий...")
    results_df = summarizer.create_csv_summary(
        vacancies,
        output_file="vacancy_summaries_demo.csv"
    )

    # Показываем статистику
    print_summary_statistics(results_df)

    # Экспортируем в Excel
    export_to_excel(results_df, "vacancy_analysis_demo.xlsx")

    # Показываем пример результата
    if len(results_df) > 0:
        print("ПРИМЕР ИЗВЛЕЧЕННЫХ ДАННЫХ:")


        sample_row = results_df.iloc[0]
        print(f"Вакансия: {sample_row['vacancy_title']}")
        print(f"Уверенность: {sample_row['confidence_score']:.2%}")
        print(f"\nТехнологии: {sample_row['extracted_technologies']}")
        print(f"\nСаммари (англ.):\n{sample_row['summary_english']}")
        print(f"\nСаммари (рус.):\n{sample_row['summary_russian']}")
        print(f"\nЗарплата: {sample_row['salary_info']}")
        print(f"Локация: {sample_row['location']}, Тип работы: {sample_row['work_type']}")


def demo_custom_vacancies():
    """Демонстрация работы с пользовательскими данными"""
    print("РАБОТА С ПОЛЬЗОВАТЕЛЬСКИМИ ДАННЫМИ")

    # Пользовательские вакансии
    custom_vacancies = [
        {
            "id": "custom_001",
            "title": "Full Stack Developer",
            "text": """Full Stack Developer with React and Node.js experience.

            Requirements:
            - 3+ years of experience with React and Node.js
            - Knowledge of TypeScript and modern JavaScript
            - Experience with MongoDB and PostgreSQL
            - Familiarity with AWS services
            - CI/CD experience with GitHub Actions

            Responsibilities:
            - Develop full-stack web applications
            - Implement responsive UI components
            - Optimize application performance
            - Write unit and integration tests

            Location: Remote
            Salary: $95,000 - $125,000
            Work type: Full-time"""
        },
        {
            "id": "custom_002",
            "title": "ML Engineer",
            "text": """Machine Learning Engineer for our AI team.

            Required:
            - Python and TensorFlow/PyTorch experience
            - 2+ years in machine learning projects
            - Knowledge of data preprocessing and model deployment
            - Experience with cloud platforms (AWS/GCP)

            Responsibilities:
            - Develop and deploy ML models
            - Optimize model performance
            - Collaborate with data scientists
            - Implement MLOps practices

            Nice to have:
            - Experience with Kubernetes
            - Knowledge of big data tools
            - Published research papers

            Location: Hybrid (London)
            Salary range: £80,000 - £110,000"""
        }
    ]

    # Обрабатываем
    summarizer = create_summarizer(verbose=True)
    results = summarizer.create_csv_summary(
        custom_vacancies,
        output_file="custom_vacancy_summaries.csv"
    )

    print_summary_statistics(results)


def show_usage_examples():
    """Показывает примеры использования модуля"""
    print("ПРИМЕРЫ ИСПОЛЬЗОВАНИЯ МОДУЛЯ")


    examples = """
    === БАЗОВОЕ ИСПОЛЬЗОВАНИЕ ===

    1. Импорт модуля:
       from vacancy_summarizer import create_summarizer
       from vacancy_data_utils import load_sample_vacancies

    2. Создание суммаризатора:
       summarizer = create_summarizer(verbose=True)

    3. Загрузка данных:
       vacancies = load_sample_vacancies()

    4. Создание CSV саммари:
       df_results = summarizer.create_csv_summary(
           vacancies,
           output_file="my_summaries.csv"
       )

    === РАБОТА С ФАЙЛАМИ ===

    1. Загрузка из CSV:
       from vacancy_data_utils import load_vacancies_from_csv
       vacancies = load_vacancies_from_csv("vacancies.csv")

    2. Загрузка из JSON:
       from vacancy_data_utils import load_vacancies_from_json
       vacancies = load_vacancies_from_json("vacancies.json")

    === АНАЛИЗ РЕЗУЛЬТАТОВ ===

    1. Статистика:
       from vacancy_data_utils import print_summary_statistics
       print_summary_statistics(df_results)

    2. Экспорт в Excel:
       from vacancy_data_utils import export_to_excel
       export_to_excel(df_results, "analysis.xlsx")

    === ПРОДВИНУТОЕ ИСПОЛЬЗОВАНИЕ ===

    1. Создание кастомных вакансий:
       custom_vacancies = [
           {
               "id": "my_vac_001",
               "title": "Job Title",
               "text": "Full job description..."
           }
       ]

    2. Пакетная обработка:
       # Автоматически создает CSV файл с результатами
       results = summarizer.create_csv_summary(vacancies)

    3. Изменение уровня детализации:
       summarizer = create_summarizer(verbose=False)  # Без вывода прогресса
    """

    print(examples)


def main():
    """Основная функция демонстрации"""
    print("МОДУЛЬ СУММАРИЗАЦИИ ВАКАНСИЙ НА АНГЛИЙСКОМ ЯЗЫКЕ")

    print("На входе: текст вакансий на английском")
    print("На выходе: CSV с аналитическим саммари")


    # Запускаем демонстрации
    demo_quick_summary()
    demo_custom_vacancies()
    show_usage_examples()

    print("ДЕМОНСТРАЦИЯ ЗАВЕРШЕНА")

    print("\nСозданные файлы:")
    print("- vacancy_summaries_demo.csv: CSV с результатами")
    print("- vacancy_analysis_demo.xlsx: Excel с аналитикой")
    print("- custom_vacancy_summaries.csv: CSV с пользовательскими данными")


if __name__ == "__main__":
    main()
```
